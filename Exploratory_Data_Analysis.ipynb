```python
def remove_outliers(data, column):
  q1 = data[column].quantile(0.25)
  q3 = data[column].quantile(0.75)
  iqr = q3 - q1

  lower_bound = q1 - 1.5 * iqr
  upper_bound = q3 + 1.5 * iqr

  outliers = data[(data[column] >= lower_bound) & (data[column] <= upper_bound)]
  return outliers
```

```python
age_outliers = remove_outliers(df, 'age')
age_outliers
```

```python
fare_outliers = remove_outliers(df, 'fare')
fare_outliers
```

```python
df['FamilySize'] = df['sibsp'] + df['parch'] + 1
```

```python
df.head()
```

```python
df["Is Alone"] = (df['FamilySize'] == 1).astype(int)
```

```python
feature = ['survived', 'pclass', 'sex', 'age', 'fare', 'embarked', 'adult_male', 'deck', 'FamilySize', 'Is Alone']
```

```python
df = df[feature]
```

```python
df['Age Group'] = pd.cut(df['age'], bins=[0, 12, 18, 35, 60, 100], labels=['Child', 'Teenager', 'Young Adult', 'Adult', 'Senior'])
```

```python
df.head()
```

```python
df['Fare Group'] = pd.cut(df['fare'], bins=[0, 10, 20, 30, 40, 50, 1000], labels=['Low', 'Medium Low', 'Medium', 'Medium High', 'High', 'Very High'])
```

```python
df.head()
```

```python
df.drop_duplicates(inplace=True)
```

```python
df.info()
```

```python
cross_table = pd.crosstab(df['sex'], df['survived'])
cross_table
```

```python
from scipy import stats
```

```python
h0 = "There is no relationship between sex and survived"
h1 = "There is a relationship between sex and survived"
chi2, p_value, dof, expected = stats.chi2_contingency(cross_table)
print("Chi-squared statistic:", chi2)
print("P-value:", p_value)

if p_value < 0.05:
  print("Reject the null hypothesis")
  print(h1)
else:
  print("Fail to reject the null hypothesis")
  print(h0)
```
